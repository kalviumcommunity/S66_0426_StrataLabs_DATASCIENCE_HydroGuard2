# Unit 4.40: Visualizing Data Distributions Using Boxplots

Use boxplots to inspect spread, median, and potential outliers across columns.

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('../data/processed/rainfall_sample_processed.csv')
df[['jun', 'jul', 'aug', 'sep', 'monsoon_total']].plot(kind='box', figsize=(8, 4))
plt.show()
```
