# Unit 4.36: Standardizing Column Names and Data Formats

Normalize column labels and text formats for consistent downstream analysis.

```python
import pandas as pd
df = pd.DataFrame({'District Name': ['Coastal North ', ' Delta West'], 'Monsoon Total': [1147, 881]})
df.columns = [c.strip().lower().replace(' ', '_') for c in df.columns]
df['district_name'] = df['district_name'].str.strip().str.lower().str.replace(' ', '_', regex=False)
df
```
