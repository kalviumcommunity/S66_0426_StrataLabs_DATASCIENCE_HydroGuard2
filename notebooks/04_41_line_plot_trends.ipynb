# Unit 4.41: Identifying Trends Over Time Using Line Plots

Use line plots to visualize year-wise trend movement in monsoon totals.

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('../data/processed/rainfall_sample_processed.csv')
trend = df.groupby('year', as_index=False)['monsoon_total'].mean()
trend.plot(x='year', y='monsoon_total', marker='o', figsize=(7, 4))
plt.show()
```
