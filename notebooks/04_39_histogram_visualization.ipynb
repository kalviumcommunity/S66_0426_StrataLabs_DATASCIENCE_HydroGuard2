# Unit 4.39: Visualizing Data Distributions Using Histograms

Use histogram plots to understand frequency distribution of numeric features.

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('../data/processed/rainfall_sample_processed.csv')
df['monsoon_total'].plot(kind='hist', bins=6, figsize=(7, 4))
plt.show()
```
